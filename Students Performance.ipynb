### SECTION B - FILTERING & CONDITIONAL SELECTION


```python
# Students Who Completed The Test Preparation Course
df[df['test preparation course'] == 'completed']
```

```python
# Students Who Scored Above 70 in Math
df[df['math score'] > 70]
```

```python
# Female Students Who Scored at Least 65 in all Three Subjects
female_high_scores = df[(df['gender'] == 'female') &
(df['math score'] >= 65) &
(df['reading score'] >= 65) &
(df['writing score'] >= 65)]
```

### SECTION C - GROUPBY & AGGREGATION

```python
# Average Math Score By Gender
df.groupby('gender')['math score'].mean()
```

```python
# Average Scores (math, reading, writing) By Parental Level of Education
df.groupby('parental level of education')[['math score', 'reading score', 'writing score']].mean()
```

```python
# Compare Average Scores Of Students With Or Without Test Preparation
df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()
```

### SECTION D - VISUALIZATION

```python
# Histogram Of Math Scores
plt.figure(figsize=(10, 6))
plt.hist(df['math score'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# Histogram Of Reading Scores
plt.figure(figsize=(10, 6))
plt.hist(df['reading score'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Reading Scores')
plt.xlabel('Reading Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# Histogram Of Writing Scores
plt.figure(figsize=(10, 6))
plt.hist(df['writing score'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Writing Scores')
plt.xlabel('Writing Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# Bar Plot Of Average Overall Score By Gender
df['average score'] = df[['math score', 'reading score', 'writing score']].mean(axis=1)
avg_score_by_gender = df.groupby('gender')['average score'].mean()
plt.figure(figsize=(10, 6))
df.groupby('gender')['average score'].mean().plot(kind='bar', color=['skyblue', 'salmon'])
plt.title('Average Score by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.show()
```

### SECTION E - INTERPRETATION

#### Does Test Preparation Appear To Improve Performance?
#### Based on the groupby comparison in Section C, students who completed the test preparation course have noticeably higher average scores across math, reading, and writing compared to those who did not. This suggests that completing the preparation course is associated with better academic performace in the dataset.
